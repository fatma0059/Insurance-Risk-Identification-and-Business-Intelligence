```python

import pymysql.cursors
import pandas as pd
model_triage = None
model_quantile = None
preprocessor = None

def find_client(identifier, prenom=None):
   
    conn = get_connection()
    if not conn:
        print(" Impossible de se connecter à la base de données")
        return None
    
    try:
        with conn.cursor() as cursor:
            if isinstance(identifier, str) and identifier.startswith('CLI-'):
                cursor.execute("""
                    SELECT id_pk, Nom, Prenom, Client_ID, Categorie_Risque 
                    FROM dim_clients 
                    WHERE Client_ID = %s
                """, (identifier,))
                client = cursor.fetchone()
                
                if client:
                    print(f" Client trouvé: {client['Nom']} {client['Prenom']} (Client_ID: {client['Client_ID']})")
                    return client['id_pk']
                else:
                    print(f" Aucun client trouvé avec le Client_ID '{identifier}'")
                    return None
            
            elif prenom:
                cursor.execute("""
                    SELECT id_pk, Nom, Prenom, Client_ID, Categorie_Risque 
                    FROM dim_clients 
                    WHERE Nom = %s AND Prenom = %s
                """, (identifier, prenom))
                client = cursor.fetchone()
                
                if client:
                    print(f" Client trouvé: {client['Nom']} {client['Prenom']} (Client_ID: {client['Client_ID']})")
                    return client['id_pk']
                else:
                    print(f" Aucun client trouvé avec le nom '{identifier} {prenom}'")
                    return None
            
            else:
                cursor.execute("""
                    SELECT id_pk, Nom, Prenom, Client_ID, Categorie_Risque 
                    FROM dim_clients 
                    WHERE Nom LIKE %s
                    ORDER BY Nom, Prenom
                """, (f"%{identifier}%",))
                clients = cursor.fetchall()
                
                if not clients:
                    print(f" Aucun client trouvé avec le nom '{identifier}'")
                    return None
                
                if len(clients) == 1:
                    client = clients[0]
                    print(f" Client trouvé: {client['Nom']} {client['Prenom']} (Client_ID: {client['Client_ID']})")
                    return client['id_pk']
                else:
                    print(f"\n Plusieurs clients trouvés avec '{identifier}':")
                    print("-" * 90)
                    print(f"{'N°':<4} | {'Client_ID':<15} | {'Nom':<15} | {'Prénom':<15} | {'Risque':<10}")
                    print("-" * 90)
                    for i, c in enumerate(clients, 1):
                        print(f"{i:<4} | {c['Client_ID']:<15} | {c['Nom']:<15} | {c['Prenom']:<15} | {c['Categorie_Risque']:<10}")
                    
                    while True:
                        try:
                            choix = input("\n Entrez le numéro du client (ou 'q' pour annuler): ")
                            if choix.lower() == 'q':
                                return None
                            
                            idx = int(choix) - 1
                            if 0 <= idx < len(clients):
                                client = clients[idx]
                                print(f" Client sélectionné: {client['Nom']} {client['Prenom']} (Client_ID: {client['Client_ID']})")
                                return client['id_pk']
                            else:
                                print(f" Choix invalide. Entrez un nombre entre 1 et {len(clients)}")
                        except ValueError:
                            print( "Veuillez entrer un nombre valide")
    
    except Exception as e:
        print(f" Erreur lors de la recherche: {e}")
        return None
    finally:
        conn.close()

def query_client(identifier, prenom=None):
    
    client_id = find_client(identifier, prenom)
    
    if not client_id:
        print(" Client non trouvé")
        return
    
    
    print(f" INFORMATIONS COMPLÈTES - CLIENT {client_id}")
   
    
    print("\n1️ INFORMATIONS GÉNÉRALES:")
    print(get_client_info(client_id))
    
    print("\n2️ PRÉDICTION DE TRIAGE:")
    print(predict_triage(client_id))
    
    print("\n3️ ESTIMATION DU COÛT:")
    print(predict_cost(client_id))

def get_client_info_by_search(identifier, prenom=None):
    client_id = find_client(identifier, prenom)
    if client_id:
        return get_client_info(client_id)
    return "Client non trouvé"

def predict_cost_by_search(identifier, prenom=None):
    client_id = find_client(identifier, prenom)
    if client_id:
        return predict_cost(client_id)
    return "Client non trouvé"

def predict_triage_by_search(identifier, prenom=None):
    client_id = find_client(identifier, prenom)
    if client_id:
        return predict_triage(client_id)
    return "Client non trouvé"

def search_and_query():
   
    print("\n" + "="*70)
    print(" RECHERCHE DE CLIENTS")
    print("\n Entrez un Client_ID (ex: CLI-21288B8C) ou un nom pour rechercher")
    print(" Tapez 'q' pour quitter")
    
    print("\n Exemples de clients disponibles:")
    conn = get_connection()
    if conn:
        try:
            with conn.cursor() as cursor:
                cursor.execute("""
                    SELECT Client_ID, Nom, Prenom 
                    FROM dim_clients 
                    LIMIT 5
                """)
                clients = cursor.fetchall()
                for c in clients:
                    print(f"  {c['Client_ID']} → {c['Nom']} {c['Prenom']}")
        finally:
            conn.close()
    
    while True:
        try:
            recherche = input("\n Recherche: ").strip()
            
            if recherche.lower() == 'q':
                print(" Au revoir!")
                break
            
            if not recherche:
                print(" Veuillez entrer une valeur")
                continue
            
            parts = recherche.split()
            
            if recherche.startswith('CLI-'):
                query_client(recherche)
            
            elif len(parts) >= 2:
                nom = parts[0]
                prenom = ' '.join(parts[1:])
                query_client(nom, prenom)
            
            else:
                query_client(parts[0])
                
        except KeyboardInterrupt:
            print("\n Au revoir!")
            break
        except Exception as e:
            print(f" Erreur: {e}")

def list_clients(limit=20):
    conn = get_connection()
    if not conn:
        print(" Impossible de se connecter")
        return
    
    try:
        with conn.cursor() as cursor:
            cursor.execute("""
                SELECT id_pk, Client_ID, Nom, Prenom, Categorie_Risque 
                FROM dim_clients 
                LIMIT %s
            """, (limit,))
            clients = cursor.fetchall()
            
            print("\n" + "="*80)
            print(f" LISTE DES CLIENTS ({len(clients)} clients)")
            print(f"{'ID':<6} | {'Client_ID':<15} | {'Nom':<15} | {'Prénom':<15} | {'Risque':<10}")
            for c in clients:
                print(f"{c['id_pk']:<6} | {c['Client_ID']:<15} | {c['Nom']:<15} | {c['Prenom']:<15} | {c['Categorie_Risque']:<10}")
    finally:
        conn.close()

def search_clients(search_term=None, limit=20):
    conn = get_connection()
    if not conn:
        print(" Impossible de se connecter")
        return
    
    try:
        with conn.cursor() as cursor:
            if search_term:
                cursor.execute("""
                    SELECT id_pk, Client_ID, Nom, Prenom, Categorie_Risque 
                    FROM dim_clients 
                    WHERE Client_ID LIKE %s 
                    OR Nom LIKE %s 
                    OR Prenom LIKE %s
                    LIMIT %s
                """, (f"%{search_term}%", f"%{search_term}%", f"%{search_term}%", limit))
            else:
                cursor.execute("""
                    SELECT id_pk, Client_ID, Nom, Prenom, Categorie_Risque 
                    FROM dim_clients 
                    LIMIT %s
                """, (limit,))
            
            clients = cursor.fetchall()
            
            if not clients:
                print(f" Aucun client trouvé pour '{search_term}'")
                return
            
            print(f" RÉSULTATS DE LA RECHERCHE ({len(clients)} clients)")
            print(f"{'N°':<4} | {'ID':<6} | {'Client_ID':<15} | {'Nom':<15} | {'Prénom':<15} | {'Risque':<10}")
            for i, c in enumerate(clients, 1):
                print(f"{i:<4} | {c['id_pk']:<6} | {c['Client_ID']:<15} | {c['Nom']:<15} | {c['Prenom']:<15} | {c['Categorie_Risque']:<10}")
    finally:
        conn.close()
```
