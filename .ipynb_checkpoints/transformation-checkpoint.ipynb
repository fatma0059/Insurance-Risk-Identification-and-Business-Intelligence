```python
import re

existants = set(dt['Immatriculation'])

def generer_nouvelle_immat(valeur, existants):
    """Modifie les 4 derniers chiffres jusqu'à obtenir une valeur unique,
    en gardant le format 123TU4567."""
    match = re.match(r'^(\d{3})TU(\d{4})$', valeur)
    if not match:
        return valeur  

    prefixe, chiffres = match.groups()
    n = int(chiffres)

    nouveau = valeur
    while nouveau in existants:
        n = (n + 1) % 10000  
        nouveau = f"{prefixe}TU{str(n).zfill(4)}"

    return nouveau

compteur = dt.groupby('Immatriculation').cumcount() + 1  
nouvelles_valeurs = []
for val, c in zip(dt['Immatriculation'], compteur):
    if c == 1:
        nouvelles_valeurs.append(val)
    else:
        nouvelle = generer_nouvelle_immat(val, existants)
        existants.add(nouvelle)
        nouvelles_valeurs.append(nouvelle)

dt['Immatriculation'] = nouvelles_valeurs


assert dt['Immatriculation'].is_unique, "Il reste des doublons !"
print("Toutes les immatriculations sont maintenant uniques .")



```

```python
def categoriser_risque(score):
    if pd.isna(score):
        return None
    elif score < 3.7:
        return 'Faible'
    elif score < 4.6:
        return 'Moyen'
    else:
        return 'Élevé'

data['Categorie_Risque'] = data['Risque_Global'].apply(categoriser_risque)
```

```python
print(data['Categorie_Risque'].value_counts())
print(data['Categorie_Risque'].value_counts(normalize=True) * 100)
```

```python

import pandas as pd

dim_temps = pd.DataFrame({'Date': pd.date_range('2020-01-01', '2028-12-31', freq='D')})


dim_temps.insert(0, 'Date_ID', range(1, len(dim_temps) + 1))


dim_temps['Date'] = dim_temps['Date'].dt.strftime('%Y-%m-%d')


assert dim_temps['Date_ID'].is_unique, "Date_ID doit être unique (clé primaire)"
assert dim_temps.isna().sum().sum() == 0, "Aucune valeur ne doit être manquante"
print(f"DimTemps générée : {dim_temps.shape[0]} jours, de {dim_temps['Date'].min()} à {dim_temps['Date'].max()}")
```

```python
FICHIER_SORTIE = "donnees_assurance_nettoyees.xlsx"
df['Date_Debut_Contrat'] = pd.to_datetime(df['Date_Debut_Contrat'], errors='coerce').dt.strftime('%Y-%m-%d')
df['Date_Fin_Contrat'] = pd.to_datetime(df['Date_Fin_Contrat'], errors='coerce').dt.strftime('%Y-%m-%d')
ds['Date_Sinistre_Claim'] = pd.to_datetime(ds['Date_Sinistre_Claim'], errors='coerce').dt.strftime('%Y-%m-%d')
assert dim_temps['Date_ID'].is_unique, "Date_ID doit être unique"
# ------------------------------------------------------------------
with pd.ExcelWriter(FICHIER_SORTIE, engine='openpyxl') as writer:
    data.to_excel(writer, sheet_name='Clients', index=False)
    df.to_excel(writer, sheet_name='Polices_Assurance', index=False)
    dt.to_excel(writer, sheet_name='Vehicules', index=False)
    ds.to_excel(writer, sheet_name='Sinistres', index=False)
    dim_temps.to_excel(writer, sheet_name='DimTemps', index=False)

print(f"Fichier final enregistré sous : {FICHIER_SORTIE}")
```
