```python
import pandas as pd
import numpy as np
import mlflow
import mlflow.sklearn
import shap
```

```python
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier, GradientBoostingRegressor
from sklearn.metrics import classification_report, f1_score, mean_pinball_loss
 
FICHIER = "donnees_assurance_nettoyees.xlsx"
```

```python
xl = pd.ExcelFile(FICHIER)
sinistres = xl.parse("Sinistres")
polices = xl.parse("Polices_Assurance")
clients = xl.parse("Clients")
vehicules = xl.parse("Vehicules")
```

```python
MOTS_GRAVES = [
    "grave",
    "majeur",
    "important",
    "multiples",
    "multiple",
    "fractures",
    "invalidité",
    "structurels",
    "structurel",
    "renversement",
    "submersion",
    "piéton",
    "cycliste",
    "car-jacking",
    "incendie",
    "vol de véhicule complet"
]
def desc_signale_gravite(description):
    return any(mot in str(description).lower() for mot in MOTS_GRAVES)
    
SEUIL_MONTANT = sinistres['Montant_Estime_Dommage_Claim'].quantile(0.75)
sinistres['Flag_Description_Grave'] = sinistres['Description_Sinistre_Claim'].apply(desc_signale_gravite)

sinistres['Cible_Triage'] = np.where(
    (sinistres['Montant_Estime_Dommage_Claim'] > SEUIL_MONTANT) |
    (sinistres['Incoherence_Dommages'] == 1) |
    (sinistres['Est_Frauduleux_Claim'] == 1) |
    (sinistres['Nature_Sinistre_Consistante'] == 0) |
    (sinistres['Flag_Description_Grave']),
    'Complexe', 'Simple'
)
print("Cible Simple/Complexe :", dict(sinistres['Cible_Triage'].value_counts()))
```

```python
df = (
    sinistres.merge(
        polices[[
            "Contract_ID",
            "Nb_Sinistres_Precedents",
            "Delai_Souscription_Sinistre_Jours",
            "Type_Couverture"
        ]],
        on="Contract_ID", how="left"
    )
    .merge(
        clients[[
            "Client_ID",
            "Age",
            "Categorie_Risque",
            "Risque_Comportemental",
            "Risque_Fraude"
        ]],
        on="Client_ID", how="left"
    )
    .merge(
        vehicules[[
            "Vehicle_ID",
            "Type_Vehicule",
            "Annee_Modele",
            "Puissance_Fiscale",
            "Kilometrage_Actuel",
            "Valeur_Vehicule",
            "Usage_Vehicule"
        ]],
        on="Vehicle_ID", how="left"
    )
)
```

```python
FEATURES_NUM = [
    "Nb_Sinistres_Precedents",
    "Delai_Souscription_Sinistre_Jours",
    "Age",
    "Risque_Comportemental",
    "Risque_Fraude",
    "Annee_Modele",
    "Puissance_Fiscale",
    "Kilometrage_Actuel",
    "Valeur_Vehicule"
]

FEATURES_CAT = [
    "Type_Sinistre_Claim",
    "Type_Couverture",
    "Categorie_Risque",
    "Type_Vehicule",
    "Usage_Vehicule"
]
```

```python
valeurs_remplacement = {c: df[c].median() for c in FEATURES_NUM}
valeurs_remplacement.update({c: "Inconnu" for c in FEATURES_CAT})

X = df[FEATURES_NUM + FEATURES_CAT].fillna(valeurs_remplacement)
y = df["Cible_Triage"]
y_bin = (y == "Complexe").astype(int)
```

```python
X_train, X_test, y_train, y_test, ybin_train, ybin_test = train_test_split(
    X, y, y_bin,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
preprocesseur = ColumnTransformer([
    ("cat", OneHotEncoder(handle_unknown="ignore"), FEATURES_CAT)
], remainder="passthrough")

```

```python
modele_rf = Pipeline([
    ("prep", preprocesseur),
    ("clf", RandomForestClassifier(
        n_estimators=300,
        max_depth=8,
        random_state=42,
        class_weight="balanced"
    ))
])

mlflow.set_experiment("triage_sinistres")

with mlflow.start_run(run_name="rf_triage_v1"):
    modele_rf.fit(X_train, y_train)
    y_pred = modele_rf.predict(X_test)

    f1_macro = f1_score(y_test, y_pred, average="macro")
    report = classification_report(y_test, y_pred, output_dict=True)

    mlflow.log_param("modele", "RandomForestClassifier")
    mlflow.log_metric("f1_macro", f1_macro)
    mlflow.log_metric("accuracy", report["accuracy"])
    mlflow.log_metric("f1_complexe", report["Complexe"]["f1-score"])

    mlflow.sklearn.log_model(modele_rf, "model")

print("\n=== RANDOM FOREST ===")
print(classification_report(y_test, y_pred))

```

```python
from sklearn.metrics import precision_recall_curve
from xgboost import XGBClassifier
poids_classe = (ybin_train == 0).sum() / (ybin_train == 1).sum()

modele_xgb = Pipeline([
    ("prep", preprocesseur),
    ("clf", XGBClassifier(
        n_estimators=400,
        max_depth=5,
        learning_rate=0.05,
        scale_pos_weight=poids_classe,
        random_state=42,
        eval_metric="logloss"
    ))
])

with mlflow.start_run(run_name="xgb_triage_v2_seuil_optimise"):
    modele_xgb.fit(X_train, ybin_train)

    proba_complexe = modele_xgb.predict_proba(X_test)[:, 1]

    precisions, rappels, seuils = precision_recall_curve(ybin_test, proba_complexe)
    f1_par_seuil = 2 * precisions * rappels / (precisions + rappels + 1e-9)

    meilleur_idx = np.argmax(f1_par_seuil[:-1])
    meilleur_seuil = seuils[meilleur_idx]

    y_pred_final = np.where(proba_complexe >= meilleur_seuil, "Complexe", "Simple")

    report_final = classification_report(y_test, y_pred_final, output_dict=True)
    f1_macro_final = f1_score(y_test, y_pred_final, average="macro")

    mlflow.log_param("modele", "XGBClassifier")
    mlflow.log_param("seuil_optimise", round(float(meilleur_seuil), 3))
    mlflow.log_metric("f1_macro", f1_macro_final)
    mlflow.log_metric("accuracy", report_final["accuracy"])
    mlflow.log_metric("f1_complexe", report_final["Complexe"]["f1-score"])

    mlflow.sklearn.log_model(modele_xgb, "model", serialization_format="pickle")

print("\n=== XGBOOST ===")
print(f"Meilleur seuil : {meilleur_seuil:.2f}")
print(classification_report(y_test, y_pred_final))
```

```python
X_test_transforme = modele_xgb.named_steps["prep"].transform(X_test)


feature_names = modele_xgb.named_steps["prep"].get_feature_names_out()


xgb_model = modele_xgb.named_steps["clf"]


explainer = shap.Explainer(xgb_model)


shap_values = explainer(X_test_transforme)
```

```python
import pandas as pd
FICHIER = "donnees_assurance_nettoyees.xlsx"

xl = pd.ExcelFile(FICHIER)
sinistres = xl.parse("Sinistres")
polices = xl.parse("Polices_Assurance")
clients = xl.parse("Clients")
vehicules = xl.parse("Vehicules")

print(f" Sinistres : {sinistres.shape[0]} lignes")
print(f" Polices : {polices.shape[0]} lignes")
print(f" Clients : {clients.shape[0]} lignes")
print(f"Véhicules : {vehicules.shape[0]} lignes")
```

```python
df = (
    sinistres.merge(
        polices[["Contract_ID", "Nb_Sinistres_Precedents", 
                 "Delai_Souscription_Sinistre_Jours", "Type_Couverture"]],
        on="Contract_ID", how="left"
    )
    .merge(
        clients[["Client_ID", "Age", "Categorie_Risque", 
                 "Risque_Comportemental", "Risque_Fraude"]],
        on="Client_ID", how="left"
    )
    .merge(
        vehicules[["Vehicle_ID", "Type_Vehicule", "Annee_Modele",
                   "Puissance_Fiscale", "Kilometrage_Actuel", 
                   "Valeur_Vehicule", "Usage_Vehicule"]],
        on="Vehicle_ID", how="left"
    )
)

FEATURES_NUM_BASE = [
    "Nb_Sinistres_Precedents", "Delai_Souscription_Sinistre_Jours",
    "Age", "Risque_Comportemental", "Risque_Fraude",
    "Annee_Modele", "Puissance_Fiscale", "Kilometrage_Actuel",
    "Valeur_Vehicule"
]

FEATURES_CAT = [
    "Type_Sinistre_Claim", "Type_Couverture", "Categorie_Risque",
    "Type_Vehicule", "Usage_Vehicule"
]


valeurs_remplacement = {c: df[c].median() for c in FEATURES_NUM_BASE}
valeurs_remplacement.update({c: "Inconnu" for c in FEATURES_CAT})

X = df[FEATURES_NUM_BASE + FEATURES_CAT].fillna(valeurs_remplacement)

print(f"✓ X créé : {X.shape[0]} lignes, {X.shape[1]} colonnes")
```

```python
print("\n AJOUT DES FEATURES AMÉLIORÉES...")

X_ameliore = X.copy()

# Features temporelles
X_ameliore['Age_Vehicule'] = 2026 - X_ameliore['Annee_Modele']
X_ameliore['Rapport_Valeur_Age'] = X_ameliore['Valeur_Vehicule'] / (X_ameliore['Age'] + 1)
X_ameliore['Kilometrage_Par_An'] = X_ameliore['Kilometrage_Actuel'] / (X_ameliore['Age_Vehicule'] + 1)
X_ameliore['Score_Risque_Global'] = (
    X_ameliore['Risque_Comportemental'] * 0.4 + 
    X_ameliore['Risque_Fraude'] * 0.3 + 
    X_ameliore['Nb_Sinistres_Precedents'] * 0.3
)

print(f" Features améliorées ajoutées : {X_ameliore.shape[1]} colonnes")
```

```python
import numpy as np
print("\n AJOUT DES FEATURES SPÉCIFIQUES Q90%...")

X_q90_ameliore = X_ameliore.copy()


X_q90_ameliore['Log_Valeur_Vehicule'] = np.log1p(X_q90_ameliore['Valeur_Vehicule'])
X_q90_ameliore['Log_Kilometrage'] = np.log1p(X_q90_ameliore['Kilometrage_Actuel'])
X_q90_ameliore['Rapport_Sinistres_An'] = X_q90_ameliore['Nb_Sinistres_Precedents'] / (X_q90_ameliore['Delai_Souscription_Sinistre_Jours'] / 365 + 1)
X_q90_ameliore['Score_Gravite_Extreme'] = (
    X_q90_ameliore['Risque_Comportemental'] * 0.5 +
    X_q90_ameliore['Risque_Fraude'] * 0.3 +
    X_q90_ameliore['Nb_Sinistres_Precedents'] * 0.2
)
X_q90_ameliore['Valeur_Risque'] = X_q90_ameliore['Valeur_Vehicule'] * X_q90_ameliore['Score_Gravite_Extreme']
X_q90_ameliore['Rapport_Valeur_Kilometrage'] = X_q90_ameliore['Valeur_Vehicule'] / (X_q90_ameliore['Kilometrage_Actuel'] + 1)
X_q90_ameliore['Puissance_par_Kilo'] = X_q90_ameliore['Puissance_Fiscale'] / (X_q90_ameliore['Kilometrage_Actuel'] + 1)


X_q90_ameliore['Poids_Q90'] = np.where(
    X_q90_ameliore['Valeur_Vehicule'] > X_q90_ameliore['Valeur_Vehicule'].quantile(0.90),
    3.0,  # Poids triple pour les très grands sinistres
    np.where(
        X_q90_ameliore['Valeur_Vehicule'] > X_q90_ameliore['Valeur_Vehicule'].quantile(0.75),
        2.0,  # Poids double pour les grands sinistres
        1.0
    )
)

NOUVELLES_FEATURES_Q90 = [
    'Log_Valeur_Vehicule', 'Log_Kilometrage', 'Rapport_Sinistres_An',
    'Score_Gravite_Extreme', 'Valeur_Risque', 'Rapport_Valeur_Kilometrage',
    'Puissance_par_Kilo', 'Poids_Q90'
]

print(f" Ajout de {len(NOUVELLES_FEATURES_Q90)} features ciblées Q90%")
print(f" Total features : {X_q90_ameliore.shape[1]}")
```

```python
class QuantileEnsemble:
    def __init__(self, quantiles=[0.1, 0.5, 0.9]):
        self.quantiles = quantiles
        self.modeles = {}
    
    def fit(self, X, y):
        for q in self.quantiles:
            print(f"  • Entraînement quantile {int(q*100)}%...")
            modele = XGBRegressor(
                objective='reg:quantileerror',
                quantile_alpha=q,
                n_estimators=300,
                max_depth=6,
                learning_rate=0.05,
                subsample=0.8,
                colsample_bytree=0.8,
                random_state=42,
                verbosity=0
            )
            modele.fit(X, y)
            self.modeles[q] = modele
        print(" Ensemble quantile entraîné")
        return self
    
    def predict(self, X):
        predictions = {}
        for q, modele in self.modeles.items():
            predictions[f'q{int(q*100)}'] = modele.predict(X)
        return predictions
    
    def evaluate(self, X, y):
        predictions = self.predict(X)
        results = {}
        for q in self.quantiles:
            key = f'q{int(q*100)}'
            pred = predictions[key]
            results[key] = {
                'mae': mean_absolute_error(y, pred),
                'rmse': np.sqrt(mean_squared_error(y, pred)),
                'r2': r2_score(y, pred),
                'pinball_loss': mean_pinball_loss(y, pred, alpha=q)
            }
        return results

print("\n Entraînement de l'ensemble quantile...")

X_cout = X.copy()
y_cout = df['Montant_Estime_Dommage_Claim'].fillna(df['Montant_Estime_Dommage_Claim'].median())

Xc_train, Xc_test, yc_train, yc_test = train_test_split(
    X_cout, y_cout, test_size=0.2, random_state=42
)

Xc_train_encoded = preprocesseur.transform(Xc_train)
Xc_test_encoded = preprocesseur.transform(Xc_test)

quantile_ensemble = QuantileEnsemble(quantiles=[0.1, 0.5, 0.9])
quantile_ensemble.fit(Xc_train_encoded, yc_train)

print("\n Évaluation de l'ensemble quantile :")
metrics_quantile = quantile_ensemble.evaluate(Xc_test_encoded, yc_test)

for q, metrics in metrics_quantile.items():
    print(f"\n  • {q}% :")
    print(f"    - MAE : {metrics['mae']:.2f}")
    print(f"    - RMSE : {metrics['rmse']:.2f}")
    print(f"    - R² : {metrics['r2']:.3f}")
    print(f"    - Pinball loss : {metrics['pinball_loss']:.2f}")
```

```python

print(" SYSTÈME MULTI-MODÈLES")

class MultiModelSystem:
    def __init__(self):
        self.modeles = {}
        self.best_model = None
        self.best_score = -np.inf
    
    def add_model(self, name, model, weight=1.0, is_binary=False):
        """
        Ajoute un modèle au système
        is_binary: True si le modèle prédit des valeurs binaires (0/1)
        """
        self.modeles[name] = {
            'model': model, 
            'weight': weight, 
            'score': None,
            'is_binary': is_binary
        }
        print(f" Modèle '{name}' ajouté")
    
    def predict_with_conversion(self, model, X, is_binary):
        """Prédit et convertit si nécessaire"""
        pred = model.predict(X)
        if is_binary:
            return np.where(pred == 1, 'Complexe', 'Simple')
        return pred
    
    def predict_proba_with_conversion(self, model, X, is_binary):
        """Prédit les probabilités et convertit si nécessaire"""
        if hasattr(model, 'predict_proba'):
            proba = model.predict_proba(X)
            if is_binary:
                
                return proba[:, 1]
            return proba[:, 1]
        return None
    
    def evaluate_model(self, name, X, y, metric='f1_macro'):
        if name not in self.modeles:
            raise ValueError(f"Modèle '{name}' non trouvé")
        
        info = self.modeles[name]
        model = info['model']
        is_binary = info['is_binary']
        
        y_pred = self.predict_with_conversion(model, X, is_binary)
        
        if metric == 'f1_macro':
            score = f1_score(y, y_pred, average='macro')
        elif metric == 'accuracy':
            score = accuracy_score(y, y_pred)
        else:
            score = model.score(X, y)
        
        self.modeles[name]['score'] = score
        if score > self.best_score:
            self.best_score = score
            self.best_model = name
        return score
    
    def evaluate_all(self, X, y, metric='f1_macro'):
        results = {}
        for name in self.modeles.keys():
            results[name] = self.evaluate_model(name, X, y, metric)
        return results
    
    def predict(self, X, method='best'):
        """Prédit avec le meilleur modèle ou avec un ensemble pondéré"""
        if method == 'best' and self.best_model:
            info = self.modeles[self.best_model]
            return self.predict_with_conversion(info['model'], X, info['is_binary'])
        elif method == 'ensemble':
            probas = []
            weights = []
            for name, info in self.modeles.items():
                if info['score'] is not None:
                    proba = self.predict_proba_with_conversion(info['model'], X, info['is_binary'])
                    if proba is not None:
                        weight = info['weight'] * (info['score'] / self.best_score)
                        probas.append(proba * weight)
                        weights.append(weight)
            
            if probas:
                ensemble_proba = np.sum(probas, axis=0) / np.sum(weights)
                return np.where(ensemble_proba >= 0.5, 'Complexe', 'Simple')
            else:
                return self.predict(X, method='best')
        else:
            first_model = list(self.modeles.values())[0]
            return self.predict_with_conversion(first_model['model'], X, first_model['is_binary'])
    
    def predict_proba(self, X):
        """Prédit les probabilités du meilleur modèle"""
        if self.best_model:
            info = self.modeles[self.best_model]
            return self.predict_proba_with_conversion(info['model'], X, info['is_binary'])
        return None

print("\n Création du système multi-modèles...")

multi_system = MultiModelSystem()

multi_system.add_model('Random Forest', modele_rf, weight=0.8, is_binary=False)
multi_system.add_model('XGBoost', modele_xgb, weight=1.0, is_binary=True)

print("\n Évaluation du système multi-modèles :")
results_multi = multi_system.evaluate_all(X_test_encoded, y_test, metric='f1_macro')

for name, score in results_multi.items():
    print(f"  • {name}: F1-macro = {score:.3f}")

print(f"\n Meilleur modèle : {multi_system.best_model} (F1-macro = {multi_system.best_score:.3f})")

print("\n Test des prédictions du système :")
y_pred_best = multi_system.predict(X_test_encoded, method='best')
print(f"  • Prédictions du meilleur modèle : {y_pred_best[:10]}")

y_pred_ensemble = multi_system.predict(X_test_encoded, method='ensemble')
print(f"  • Prédictions de l'ensemble : {y_pred_ensemble[:10]}")

print("\n Comparaison des prédictions :")
comparaison_pred = pd.DataFrame({
    'Réel': y_test[:10].values,
    'Meilleur_modèle': y_pred_best[:10],
    'Ensemble': y_pred_ensemble[:10]
})
print(comparaison_pred.to_string(index=False))
```

```python
print(" SIMULATION DES COÛTS SUR 5 ANS")
class CostSimulator:
    def __init__(self, model_quantile, inflation_rate=0.02, annual_claims_growth=0.01):
        self.model_quantile = model_quantile
        self.inflation_rate = inflation_rate
        self.annual_claims_growth = annual_claims_growth
        self.historique = []
    
    def simulate_year(self, X_base, year, n_samples=1000):
        n_claims = int(len(X_base) * (1 + self.annual_claims_growth) ** year)
        idx = np.random.choice(len(X_base), n_claims, replace=True)
        X_sample = X_base.iloc[idx].copy()
        X_sample_encoded = preprocesseur.transform(X_sample)
        predictions = self.model_quantile.predict(X_sample_encoded)
        inflation_factor = (1 + self.inflation_rate) ** year
        for key in predictions:
            predictions[key] = predictions[key] * inflation_factor
        
        return {
            'year': 2026 + year,
            'n_claims': n_claims,
            'mean_q50': predictions['q50'].mean(),
            'total_cost_lower': predictions['q10'].sum(),
            'total_cost_median': predictions['q50'].sum(),
            'total_cost_upper': predictions['q90'].sum(),
            'inflation_factor': inflation_factor
        }
    
    def simulate(self, X_base, years=5, n_samples=1000):
        self.historique = []
        for year in range(years):
            self.historique.append(self.simulate_year(X_base, year, n_samples))
        return pd.DataFrame(self.historique)
    
    def plot_simulation(self):
        df = pd.DataFrame(self.historique)
        fig, axes = plt.subplots(2, 1, figsize=(12, 10))
        
        axes[0].plot(df['year'], df['total_cost_median']/1e6, 'b-', label='Coût médian', linewidth=2)
        axes[0].fill_between(df['year'], df['total_cost_lower']/1e6, df['total_cost_upper']/1e6, alpha=0.3, color='blue', label='Intervalle 10%-90%')
        axes[0].set_xlabel('Année')
        axes[0].set_ylabel('Coût total (millions €)')
        axes[0].set_title('Évolution des coûts totaux', fontsize=12, fontweight='bold')
        axes[0].legend()
        axes[0].grid(True, alpha=0.3)
        
        axes[1].bar(df['year'], df['n_claims'], color='#3498db', alpha=0.7)
        axes[1].set_xlabel('Année')
        axes[1].set_ylabel('Nombre de sinistres')
        axes[1].set_title('Évolution du nombre de sinistres', fontsize=12, fontweight='bold')
        axes[1].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.savefig('simulation_couts_5ans.png', dpi=300, bbox_inches='tight')
        plt.show()
        plt.close()
        print(" Graphique sauvegardé : simulation_couts_5ans.png")

simulator = CostSimulator(model_quantile=quantile_ensemble, inflation_rate=0.02, annual_claims_growth=0.01)
simulation_results = simulator.simulate(X, years=5, n_samples=2000)

print("\n Résultats de la simulation :")
print(simulation_results[['year', 'n_claims', 'total_cost_lower', 'total_cost_median', 'total_cost_upper']].round(2))
simulator.plot_simulation()
```
